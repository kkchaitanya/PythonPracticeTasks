```python
from qdrant_client.models import PointStruct
points = []
for chunk,vectors in zip(chunks,vectors):
    points.append(PointStruct(
        id=chunk["chunk_id"],
        vector=vectors,
        payload={
            "text": chunk["text"],
            "topic": chunk["topic"],
            "chunk_id": chunk["chunk_id"],
            "source": "super_30"
        }
    ))  # Closing PointStruct and append
    # points.append(point)

qdrant.upsert(
    collection_name=collection_name,
    points=points
)
```

```python
query =" What is Euron and what are all the technologies and features that they have? What are the products that they have?"
```

```python
resp = client.embeddings.create(
    model="gemini-embedding-2-preview",
    input=query,
)
```

```python
result = qdrant.query_points(
    collection_name=collection_name,
    query=resp.data[0].embedding,
    limit=2,
    with_payload=True
)
```

```python
retrieved_context = "\n\n".join(
    result.payload["text"]
    for result in result.points
)
```

```python
response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {
            "role": "system",
            "content": """
You are a helpful RAG assistant.

Answer the user's question using only the
retrieved context.

Rules:
1. Do not invent information.
2. Keep the answer clear and accurate.
3. If the information is missing, say
4. always try to check the name that have been given in query and in retrival
   "I don't know based on the provided context."
5. Always give me clean data. Remove all the statics, `\n`, tabs, and all the fillers.
"""
        },
        {
            "role": "user",
            "content": f"""
Retrieved Context:
{retrieved_context}

Question:
{query}
"""
        }
    ]
)

final_answer = response.choices[0].message.content
```
