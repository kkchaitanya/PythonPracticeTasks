```python
l = [
    "apple",
    "banana",
    "avocado",
    "blueberry",
    "cherry",
    "carrot"
]
d = {}
for i in l:
    if i[0] not in list(d.keys()):
        d[i[0]] = [i]
    else:
        d[i[0]].append(i)
    
print(d)
```
