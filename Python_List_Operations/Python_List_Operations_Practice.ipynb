Combining lists – RAG retrieval results

```python
documents = ["Doc1", "Doc2", "Doc3"]
scores = [0.95, 0.87, 0.76]
```

```python
for doc,sc in zip(documents,scores):
    print(doc)
    print(sc)
```

```python
retrieved_docs = [
    "Python supports list operations",
    "Python supports dictionary operations",
    "RAG uses document retrieval",
    "LangChain supports retrievers"
]
scores = [0.96, 0.89, 0.92, 0.81]
```

```python
for index, (rt,sc) in enumerate(zip(retrieved_docs,scores)):
    print(index)
    print(rt)
    print(sc)
```

```python
for index, data in enumerate(retrieved_docs):
    print(index)
    print(data)
```
